```python
# Q1. Find the Shortest Word in a Sentence
# Logic (Markdown cell):
# Observe: A sentence contains multiple words.
# Understand: Find the word with the minimum length.
# Break: Split the sentence into words.
# Logic: Assume the first word is shortest and compare the remaining words.
# Code: Use a for loop.
# Test: Check the example and another sentence.

sentence = input("Enter a sentence: ")
words = sentence.split()
shortest = words[0]
for word in words:
    if len(word) < len(shortest):
        shortest = word
print(shortest)
```

```python
# Q2. Find the Largest Number in a List
# Logic: Start with the first number. Compare every number with the current largest number and update it whenever a larger number is found.

numbers = [12, 45, 7, 89, 23]
largest = numbers[0]
for num in numbers:
    if num > largest:
        largest = num
print(largest)
# Additional test
numbers = [5, 15, 2]
largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num

print(largest)
```

```python
# Q3. Find the Smallest Number in a List
# Logic: Start with the first number. Compare the remaining numbers and update the smallest whenever a smaller number is found.

numbers = [12, 45, 7, 89, 23]

smallest = numbers[0]

for num in numbers:
    if num < smallest:
        smallest = num

print(smallest)

# Additional test
numbers = [8, 3, 10]
smallest = numbers[0]
for num in numbers:
    if num < smallest:
        smallest = num

print(smallest)
```

```python
# Q4. Find the First Longest Word When Ties Occur
# Logic: Start with the first word. Update it only when a strictly longer word appears. This preserves the first word when lengths are equal.

sentence = input("Enter a sentence: ")
words = sentence.split()

longest = words[0]

for word in words:
    if len(word) > len(longest):
        longest = word

print(longest)
```

```python
# Q5. Find the Word with the Maximum Number of Characters
# Logic: Split the sentence into words, compare their lengths, and keep the longest word found.

sentence = input("Enter a sentence: ")
words = sentence.split()

result = words[0]

for word in words:
    if len(word) > len(result):
        result = word

print(result)
```

```python
# Q6. Count Vowels in a String
# Logic: Check each character. If it belongs to aeiouAEIOU, increase the counter.

text = input("Enter a string: ")
count = 0

for char in text:
    if char in "aeiouAEIOU":
        count += 1

print(count)
```

```python
# Q7. Count Consonants in a String
# Logic: Count characters that are alphabetic but are not vowels. Spaces, digits, and special characters are ignored.

text = input("Enter a string: ")
count = 0

for char in text:
    if char.isalpha() and char.lower() not in "aeiou":
        count += 1

print(count)
```

```python
# Q8. Find the Most Frequent Character
# Logic: Ignore spaces, count each character using a dictionary, and find the character with the highest frequency. In a tie, keep the character encountered firs

text = input("Enter a string: ")
frequency = {}

for char in text:
    if char != " ":
        frequency[char] = frequency.get(char, 0) + 1

result = ""
highest = 0

for char in frequency:
    if frequency[char] > highest:
        highest = frequency[char]
        result = char

print(result)
```

```python
# Q9. Find the Least Frequent Character
# Logic: Count the frequency of each non-space character. Find the smallest frequency and preserve the first character when frequencies are tied.

text = input("Enter a string: ")
frequency = {}

for char in text:
    if char != " ":
        frequency[char] = frequency.get(char, 0) + 1

result = ""
lowest = len(text) + 1

for char in frequency:
    if frequency[char] < lowest:
        lowest = frequency[char]
        result = char

print(result)
```

```python
# Q10. Find All Characters Appearing More Than Once
# Logic: Count each non-space character. Then collect characters whose count is greater than one, displaying each only once in its first-appearance order.

text = input("Enter a string: ")
frequency = {}

for char in text:
    if char != " ":
        frequency[char] = frequency.get(char, 0) + 1

result = []

for char in text:
    if char != " " and frequency[char] > 1 and char not in result:
        result.append(char)

print(result)
```

```python
# Q11. Dictionary of Words and Their First Characters
# Logic: Split the sentence into words. Store each word as a key and its first character as the value.

sentence = input("Enter a sentence: ")
words = sentence.split()
result = {}

for word in words:
    result[word] = word[0]

print(result)
```

```python
# Q12. Dictionary of Words and Their Last Characters
# Logic: Use word[-1] to access the last character of each word.

sentence = input("Enter a sentence: ")
words = sentence.split()
result = {}

for word in words:
    result[word] = word[-1]

print(result)
```

```python
# Q13. Dictionary of Words and Their Uppercase Versions
# Logic: Use each original word as the key and convert it to uppercase for the value.

sentence = input("Enter a sentence: ")
words = sentence.split()
result = {}

for word in words:
    result[word] = word.upper()

print(result)
```

```python
# Q15. Dictionary of Numbers and Their Even/Odd Status
# Logic: If a number is divisible by 2, store "Even"; otherwise, store "Odd".

numbers = [1, 2, 3, 4, 5]
result = {}

for num in numbers:
    if num % 2 == 0:
        result[num] = "Even"
    else:
        result[num] = "Odd"

print(result)

# Additional test
numbers = [6, 7]
result = {}

for num in numbers:
    if num % 2 == 0:
        result[num] = "Even"
    else:
        result[num] = "Odd"

print(result)
```

```python
# Q16. Find the Most Frequent Word in a Sentence
# Logic: Convert words to lowercase, count each word, and return the first word with the highest frequency.

sentence = input("Enter a sentence: ")
words = sentence.lower().split()
frequency = {}

for word in words:
    frequency[word] = frequency.get(word, 0) + 1

result = ""
highest = 0

for word in words:
    if frequency[word] > highest:
        highest = frequency[word]
        result = word

print(result)
```

```python
# Q17. Find All Words Appearing More Than Once
# Logic: Convert words to lowercase, count their occurrences, and collect repeated words once in first-appearance order.

sentence = input("Enter a sentence: ")
words = sentence.lower().split()
frequency = {}

for word in words:
    frequency[word] = frequency.get(word, 0) + 1

result = []

for word in words:
    if frequency[word] > 1 and word not in result:
        result.append(word)

print(result)

```

```python
# Q18. Find the First Word That Appears Only Once
# Logic: Count the lowercase words, then traverse the original word order and select the first word with frequency one.

sentence = input("Enter a sentence: ")
words = sentence.lower().split()
frequency = {}

for word in words:
    frequency[word] = frequency.get(word, 0) + 1

result = None

for word in words:
    if frequency[word] == 1:
        result = word
        break

print(result)
```

```python
# Q19. Count Words Whose Length Is Greater Than 4
# Logic: Split the sentence into words and increase the counter whenever a word's length is greater than four.

sentence = input("Enter a sentence: ")
words = sentence.split()
count = 0

for word in words:
    if len(word) > 4:
        count += 1

print(count)
```

```python
# Q20. Find the Longest Word That Appears Only Once
# Logic: Convert words to lowercase, count their frequencies, and find the longest word occurring exactly once. Update only when a strictly longer word is found, preserving the first word in a tie.

sentence = input("Enter a sentence: ")
words = sentence.lower().split()
frequency = {}

for word in words:
    frequency[word] = frequency.get(word, 0) + 1

result = ""
longest = 0

for word in words:
    if frequency[word] == 1 and len(word) > longest:
        longest = len(word)
        result = word

print(result if result else None)

```
