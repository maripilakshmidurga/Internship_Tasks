## Q1. Find the Frequency of Each Word in a Sentence

### Pseudocode
1. Create an empty dictionary.
2. Split the sentence into words.
3. Visit each word using a loop.
4. If the word exists, increase its count.
5. Otherwise, add the word with count 1.
6. Return the dictionary.


```python

def word_frequency(sentence):
    result = {}

    for word in sentence.split():
        if word in result:
            result[word] += 1
        else:
            result[word] = 1

    return result


print(word_frequency("python is easy and python is powerful"))
print(word_frequency("hello hello world"))
```

## Q2. Find the Frequency of Each Character in a Word

### Pseudocode
1. Create an empty dictionary.
2. Read each character in the word.
3. If the character exists, increase its count.
4. Otherwise, store the character with count 1.
5. Return the dictionary.

```python
def character_frequency(word):
    result = {}
    for char in word:
        if char in result:
            result[char] += 1
        else:
            result[char] = 1
    return result

print(character_frequency("banana"))
print(character_frequency("hello"))
```

## Q3. Find the First Non-Repeating Character

### Pseudocode
1. Read each character from left to right.
2. Count how many times the character appears.
3. If its count is 1, return that character.
4. If no such character exists, return None.

```python
def first_non_repeating(word):
    for char in word:
        if word.count(char) == 1:
            return char
    return None

print(first_non_repeating("swiss"))
print(first_non_repeating("aabb"))
```

## Q4. Check Whether Two Strings Are Anagrams

### Pseudocode
1. Sort the characters of both strings.
2. Compare the sorted results.
3. Return True if they match; otherwise, return False.

```python
def are_anagrams(s1, s2):
    return sorted(s1) == sorted(s2)

print(are_anagrams("listen", "silent"))
print(are_anagrams("hello", "world"))
```

## Q5. Find the First Repeating Character

### Pseudocode
1. Create an empty set to store visited characters.
2. Read each character from left to right.
3. If the character is already in the set, return it.
4. Otherwise, add it to the set.
5. If no character repeats, return None.

```python
def first_repeating_char(word):
    seen = set()
    for char in word:
        if char in seen:
            return char
        seen.add(char)
    return None

print(first_repeating_char("programming"))
print(first_repeating_char("python"))
```

## Q6. Move All Zeros to the End

### Pseudocode
1. Collect all non-zero numbers in their original order.
2. Count the zeros in the original list.
3. Add the zeros at the end.
4. Return the new list.

```python
def move_zeros(nums):
    result = [num for num in nums if num != 0]
    result.extend([0] * (len(nums) - len(result)))
    return result

print(move_zeros([0, 3, 0, 1, 2, 0, 4]))
print(move_zeros([0, 1, 0, 2]))
```

## Q7. Find the Second Highest Distinct Number

### Pseudocode
1. Remove duplicate numbers.
2. Sort the unique numbers in descending order.
3. If fewer than two unique numbers exist, return None.
4. Otherwise, return the second number.

```python
def second_highest(nums):
    unique_nums = sorted(set(nums), reverse=True)
    if len(unique_nums) < 2:
        return None
    return unique_nums[1]

print(second_highest([10, 20, 40, 30, 40, 20]))
print(second_highest([5, 5, 5]))
```

## Q8. Remove Duplicates Without Changing Order

### Pseudocode
1. Create an empty result list.
2. Visit each element in the original list.
3. Add it only if it is not already in the result.
4. Return the result list.

```python
def remove_duplicates(nums):
    result = []
    for num in nums:
        if num not in result:
            result.append(num)
    return result

print(remove_duplicates([4, 2, 4, 5, 2, 7, 5]))
print(remove_duplicates([1, 1, 2, 3, 2]))
```

## Q9. Find the Missing Number from 1 to N

### Pseudocode
1. Calculate the expected sum from 1 to N.
2. Calculate the sum of the given list.
3. Subtract the list sum from the expected sum.
4. Return the difference.

```python
def missing_number(nums, n):
    expected_sum = n * (n + 1) // 2
    return expected_sum - sum(nums)

print(missing_number([1, 2, 4, 5, 6], 6))
print(missing_number([1, 2, 3, 5], 5))
```

## Q10. Find Two Numbers Whose Sum Equals a Target

### Pseudocode
1. Create an empty list of previously visited numbers.
2. For each number, calculate the required partner.
3. If the partner was already visited, return the pair.
4. Otherwise, store the current number.
5. If no pair is found, return an empty list.

```python
def two_sum(nums, target):
    seen = []
    for num in nums:
        needed = target - num
        if needed in seen:
            return [needed, num]
        seen.append(num)
    return []

print(two_sum([2, 7, 11, 15], 9))
print(two_sum([1, 4, 6, 8], 10))
```

## Q11. Find the Number Closest to Zero

### Pseudocode
1. Compare numbers using their absolute values.
2. Select the number with the smallest absolute value.
3. If tied, choose the positive number.
4. Return the selected number.

```python
def closest_to_zero(nums):
    return min(nums, key=lambda num: (abs(num), -num))

print(closest_to_zero([-8, -3, 2, 5, -2]))
print(closest_to_zero([-4, -1, 1, 3]))
```

## Q12. Find Common Elements Between Two Lists

### Pseudocode
1. Create an empty result list.
2. Visit elements in the first list.
3. Check whether each element exists in the second list.
4. Add it if it is common and not already in the result.
5. Return the result.

```python
def common_elements(list1, list2):
    result = []
    for num in list1:
        if num in list2 and num not in result:
            result.append(num)
    return result

print(common_elements([1, 2, 3, 4, 5], [3, 4, 5, 6, 7]))
print(common_elements([2, 2, 3, 4], [2, 4, 6]))
```

## Q13. Rotate a List to the Right by K Positions

### Pseudocode
1. If the list is empty, return an empty list.
2. Reduce K using the list length.
3. Move the last K elements to the beginning.
4. Add the remaining elements after them.
5. Return the rotated list.

```python
def rotate_right(nums, k):
    if not nums:
        return []
    k = k % len(nums)
    return nums[-k:] + nums[:-k] if k else nums[:]

print(rotate_right([1, 2, 3, 4, 5], 2))
print(rotate_right([1, 2, 3, 4], 1))
```

## Q14. Find the Maximum Sum of Two Consecutive Elements

### Pseudocode
1. Calculate the sum of the first two elements.
2. Visit every remaining consecutive pair.
3. Calculate each pair's sum.
4. Update the maximum when a larger sum is found.
5. Return the maximum sum.

```python
def max_consecutive_sum(nums):
    max_sum = nums[0] + nums[1]
    for i in range(1, len(nums) - 1):
        current_sum = nums[i] + nums[i + 1]
        if current_sum > max_sum:
            max_sum = current_sum
    return max_sum

print(max_consecutive_sum([4, 7, 2, 9, 5]))
print(max_consecutive_sum([1, 8, 3, 6]))
```

## Q15. Find All Pairs of Numbers Whose Sum Equals a Target

### Pseudocode
1. Create an empty result list and a set of visited numbers.
2. For each number, calculate its required partner.
3. If the partner was visited, create the pair in ascending order.
4. Add the pair only if it is not already present.
5. Sort the pairs by their first element.
6. Return the result list.

```python
def all_pairs(nums, target):
    result = []
    seen = set()

    for num in nums:
        needed = target - num
        if needed in seen:
            pair = sorted([needed, num])
            if pair not in result:
                result.append(pair)
        seen.add(num)

    result.sort(key=lambda pair: pair[0])
    return result

print(all_pairs([2, 4, 3, 5, 7, 8, 1], 9))
print(all_pairs([1, 2, 3, 4, 5], 6))
```
